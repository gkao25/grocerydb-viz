When I try to compare Target, Walmart, and WholeFoods, the first difference that come to mind is the price different and how they're generally viewed as different tiers of grocery stores. WholeFoods is often considered to be one of the most expensive grocery stores, and I want to prove this by using the `grocerydb` dataset. I aggregated the dataset by store and brand to get the average price per brand at differnt stores. The first plot was an overlapping barplot, so the tallest bar for a specific brand would mean that this brand is most expensive at X store. I felt like bars make it unclear which store has the highest number of "the tallest bar" without a table with explicit numbers, so I made a new boxplot, which can show the center statistic and spread clearly. Now, it's easier to see at a first glance that for the same brand, WholeFoods sells it at a higher price. I also wonder if aggregating the items by brand was necessary, so there's another boxplot (comparing different store) just by plotting the raw prices. However, this seems too cluttered with many points outside the maximum whisker. In conclusion, I am most satisfied with my second plot.
